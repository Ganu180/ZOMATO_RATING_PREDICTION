```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import GradientBoostingRegressor
print("✅ Tuning libraries imported successfully!")
```

```python
if best_model_name == "Random Forest":
    selected_model = RandomForestRegressor(random_state=42,n_jobs=-1)
elif best_model_name == "Linear Regression":
    selected_model = LinearRegression()
elif best_model_name == "Gradient Boosting":
    selected_model = GradientBoostingRegressor(random_state=42)
else:
    raise ValueError(f"Unknown model: {best_model_name}")
print(f"✅ Selected model: {best_model_name}")
```

```python
if best_model_name == "Random Forest":
    param_grid = {
        "n_estimators": [100, 200, 300],
        "max_depth": [10, 15, 20, None],
        "min_samples_split": [2, 5, 10],
        "min_samples_leaf": [1, 2, 4],
        "max_features": ["sqrt", "log2"]
    }
elif best_model_name == "Gradient Boosting":
    param_grid = {
        "n_estimators": [100, 200],
        "learning_rate": [0.03, 0.05, 0.1],
        "max_depth": [3, 5],
        "min_samples_split": [2, 5],
        "min_samples_leaf": [1, 2]
    }
elif best_model_name == "Linear Regression":
    param_grid = {}
print("Hyperparameter grid created successfully!")
for parameter, values in param_grid.items():
    print(f"{parameter}: {values}")
```

```python
from sklearn.model_selection import RandomizedSearchCV
if best_model_name == "Linear Regression":
    tuned_model = selected_model
    tuned_model.fit(X_train_processed, y_train)
    best_params = {}
    print("Linear Regression does not require hyperparameter tuning.")
else:
    random_search = RandomizedSearchCV(
        estimator=selected_model,
        param_distributions=param_grid,
        n_iter=20,
        cv=5,
        scoring="r2",
        random_state=42,
        n_jobs=-1,
        verbose=2
    )
    random_search.fit(X_train_processed,y_train)
    tuned_model = random_search.best_estimator_
    best_params = random_search.best_params_
    print("\n✅ Hyperparameter tuning completed!")
    print("\nBest Cross-Validation R²:")
    print(round(random_search.best_score_, 6))
print("\nBest Parameters:")
print(best_params)
```

```python
print("=" * 60)
print("BEST HYPERPARAMETERS")
print("=" * 60)
print(f"Model: {best_model_name}")
if best_params:
    for parameter, value in best_params.items():
        print(f"{parameter}: {value}")
else:
    print("No tunable hyperparameters.")
```

```python
def predict_restaurant_rating(
    online_order,
    book_table,
    votes,
    location,
    rest_type,
    cuisines,
    approx_costfor_two_people,
    listed_intype,
    listed_incity
):
    new_restaurant = pd.DataFrame({
        "online_order": [online_order],
        "book_table": [book_table],
        "votes": [votes],
        "location": [location],
        "rest_type": [rest_type],
        "cuisines": [cuisines],
        "approx_costfor_two_people": [approx_costfor_two_people],
        "listed_intype": [listed_intype],
        "listed_incity": [listed_incity]
    })
    processed_input = loaded_preprocessor.transform(new_restaurant)
    prediction = loaded_model.predict(processed_input)[0]
    prediction = np.clip(
        prediction,
        1.0,
        5.0
    )
    return prediction
```

```python
def restaurant_rating_predictor(
    online_order,
    book_table,
    votes,
    location,
    rest_type,
    cuisines,
    cost,
    listed_type,
    city
):
    online_value = 1 if online_order.lower() == "yes" else 0
    book_value = 1 if book_table.lower() == "yes" else 0
    prediction = predict_restaurant_rating(
        online_order=online_value,
        book_table=book_value,
        votes=votes,
        location=location,
        rest_type=rest_type,
        cuisines=cuisines,
        approx_costfor_two_people=cost,
        listed_intype=listed_type,
        listed_incity=city
    )
    print("=" * 60)
    print("ZOMATO RESTAURANT RATING PREDICTOR")
    print("=" * 60)
    print(f"Online Order : {online_order}")
    print(f"Book Table   : {book_table}")
    print(f"Votes        : {votes}")
    print(f"Location     : {location}")
    print(f"Cuisine      : {cuisines}")
    print(f"Cost for Two : ₹{cost}")
    print("-" * 60)
    print(f"Predicted Rating: {prediction:.2f} / 5")
```
